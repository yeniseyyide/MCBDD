# Factor Analysis Tutorial

This tutorial introduces the concept of factor analysis and demonstrates how to apply it using Python.

## 1. Introduction

Factor Analysis is a statistical method used to describe variability among observed variables in terms of fewer unobserved variables called *factors*. 

The idea is that multiple observed variables may be influenced by a smaller number of underlying latent factors.

For example, in psychology, many survey questions might measure only a few underlying personality traits.

## 2. Intuition

Imagine you have a questionnaire with 10 questions about personality.

Some questions might relate to sociability, while others relate to emotional stability.

Instead of analyzing all 10 questions separately, factor analysis tries to identify a smaller number of hidden variables (factors), such as:
- Extraversion
- Neuroticism

This reduces complexity and helps us understand the structure of the data.

## 3. Mathematical Idea

Factor analysis assumes that each observed variable is a linear combination of latent factors plus some noise.

X = LF + ε

## 4. Use Case

We simulate data where multiple observed variables depend on two hidden factors and try to recover them.

```python
import numpy as np
import pandas as pd
from sklearn.decomposition import FactorAnalysis
import matplotlib.pyplot as plt

np.random.seed(42)

n_samples = 200
factor1 = np.random.normal(size=n_samples)
factor2 = np.random.normal(size=n_samples)

X = np.array([
    0.8 * factor1 + 0.2 * factor2 + np.random.normal(scale=0.1, size=n_samples),
    0.7 * factor1 + 0.1 * factor2 + np.random.normal(scale=0.1, size=n_samples),
    0.2 * factor1 + 0.9 * factor2 + np.random.normal(scale=0.1, size=n_samples),
    0.1 * factor1 + 0.8 * factor2 + np.random.normal(scale=0.1, size=n_samples),
    0.5 * factor1 + 0.5 * factor2 + np.random.normal(scale=0.1, size=n_samples)
]).T

fa = FactorAnalysis(n_components=2)
X_factors = fa.fit_transform(X)

df_factors = pd.DataFrame(X_factors, columns=["Factor 1", "Factor 2"])
df_factors.head()
```

```python
plt.scatter(df_factors["Factor 1"], df_factors["Factor 2"])
plt.title("Recovered Factors")
plt.xlabel("Factor 1")
plt.ylabel("Factor 2")
plt.show()
```

## 5. Interpretation

The factor analysis reduces the dataset into two dimensions and captures the main structure of the data.

## 6. Conclusion

Factor analysis helps uncover hidden structure and reduce dimensionality in datasets.
